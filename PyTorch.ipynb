# **Import PyTorch**

```python
# Import PyTorch library
import torch

# Import Neural Network module
import torch.nn as nn

# Import Optimization module
import torch.optim as optim
```

# **Create Tensors**
**torch.tensor()**

```python
# Create a tensor

tensor = torch.tensor([10,20,30,40])

print(tensor)
```

**torch.zeros()**

```python
# Create a tensor filled with zeros

tensor = torch.zeros(2,3)

print(tensor)
```

**torch.ones()**

```python
# Create a tensor filled with ones

tensor = torch.ones(2,3)

print(tensor)
```

**torch.rand()**

```python
# Create a tensor with random values

tensor = torch.rand(2,3)

print(tensor)
```

**torch.randn()**

```python
# Create a tensor with random values

tensor = torch.rand(2,3)

print(tensor)
```

# **Tensor Operations**
**Addition**

```python
a = torch.tensor([1,2,3])
b = torch.tensor([4,5,6])

print(a+b)
```

**Subtraction**

```python
print(a-b)
```

**Multiplication**

```python
print(a*b)
```

**Division**

```python
print(a/b)
```

**Matrix Multiplication**

```python
A = torch.tensor([[1,2],[3,4]])

B = torch.tensor([[5,6],[7,8]])

print(torch.matmul(A,B))
```

# **GPU**

# **Automatic Differentiation**
**requires_grad**

```python
x = torch.tensor(2.0,requires_grad=True)

print(x)
```

**backward()**

```python
y = x**2

y.backward()

print(x.grad)
```

# **Neural Networks**
**nn.Linear()**

```python
model = nn.Linear(2,1)

print(model)
```

**nn.ReLU()**

```python
relu = nn.ReLU()

x = torch.tensor([-1.0,0.0,2.0])

print(relu(x))
```

**nn.Sequential()**

```python
model = nn.Sequential(
    nn.Linear(2,4),
    nn.ReLU(),
    nn.Linear(4,1)
)

print(model)
```

# **Loss Functions**
**nn.MSELoss()**

```python
loss_function = nn.MSELoss()

print(loss_function)
```

**nn.CrossEntropyLoss()**

```python
loss_function = nn.CrossEntropyLoss()

print(loss_function)
```

# **Optimizers**
**SGD**

```python
optimizer = optim.SGD(
    model.parameters(),
    lr=0.01
)

print(optimizer)
```

**Adam**

```python
optimizer = optim.Adam(
    model.parameters(),
    lr=0.001
)

print(optimizer)
```

# **Training Loop**

```python
# Forward Pass

output = model(torch.rand(1,2))

# Calculate Loss

loss = nn.MSELoss()(output,torch.rand(1,1))

# Clear previous gradients

optimizer.zero_grad()

# Backpropagation

loss.backward()

# Update weights

optimizer.step()
```

# **Save Model**
**torch.save()**

```python
torch.save(model.state_dict(),"model.pth")
```

**torch.load(**)

```python
model.load_state_dict(torch.load("model.pth"))
```
