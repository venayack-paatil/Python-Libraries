## Cell 2 – Import Keras"Importing" a library means telling Python: *"I want to use the tools inside this toolbox."*- `tensorflow` → the big toolbox that contains Keras.- `keras` → the beginner-friendly part of TensorFlow we'll use to build our model.- `numpy` → helps us create and work with arrays (lists of numbers) efficiently.- `matplotlib.pyplot` → helps us draw graphs and charts.

## Cell 3 – Purpose of Keras### DefinitionKeras is a **high-level Deep Learning API** built into TensorFlow."High-level" means it hides the complicated math and lets us build models using simple, readable code — like giving instructions in plain English instead of complex formulas.### Why Keras?- ✅ Easy to build Deep Learning models- ✅ Beginner friendly- ✅ Less code than raw TensorFlow- ✅ Supports CNNs (for images), RNNs/LSTMs (for sequences like text), and Transfer Learning (reusing already-trained models)### Importance- Simple syntax (easy to read and write)- Fast development (build models quickly)- Industry standard, used together with TensorFlow at real companies- Works on CPU, GPU, and TPU (different types of computer "brains" that can run the model, some much faster than others)### Real-world Applications- 🖼️ Image Classification (e.g., "is this a cat or a dog?")- 🙂 Face Recognition- 💬 Chatbots- 🏥 Medical Diagnosis- 🎬 Recommendation Systems (like "movies you might like")

## Cell 4 – Create Sample DatasetA **dataset** is just a collection of examples we use to teach our model.Here, we're teaching the model a simple pattern: **y = 2 × x**- `X` is our **input** (the question).- `y` is our **output** (the correct answer).Notice the pattern: 1→2, 2→4, 3→6, 4→8, 5→10. Each `y` value is double the `X` value. Our goal is for the model to *discover* this pattern on its own, just by looking at examples!

```python
# Create our input values (X) as a 2D array.# Each number is inside its own [ ] because Keras expects each example# to be a "row" of features, even if there's only 1 feature.X = np.array([[1], [2], [3], [4], [5]], dtype=np.float32)# Create the matching output values (y). Notice y = 2 * X for every row.y = np.array([[2], [4], [6], [8], [10]], dtype=np.float32)# Let's print them out to see what they look likeprint("Inputs (X):")print(X)print("Outputs (y):")print(y)
```

## Cell 5 – Create Sequential ModelA **Sequential model** is the simplest type of Keras model — it's just a straight stack of layers, one after another, like a sandwich 🥪.- `keras.Input(shape=(1,))` → tells the model: "each example has exactly 1 number as input" (our `X` value).- `keras.layers.Dense(8, activation="relu")` → a **Dense layer** with 8 "neurons." Each neuron is like a tiny calculator that looks for patterns.  - `activation="relu"` adds a bit of "bendiness" to help the model learn more complex patterns (without it, the model could only learn straight lines).- `keras.layers.Dense(1)` → the final layer with 1 neuron, giving us **one output number** — our prediction.`model.summary()` prints a table showing the structure of our model — like a blueprint.

```python
# Build a simple Sequential (stack of layers) modelmodel = keras.Sequential([    keras.Input(shape=(1,)),                  # 1 input number    keras.layers.Dense(8, activation="relu"),  # hidden layer with 8 neurons    keras.layers.Dense(1)                      # output layer with 1 number])# Show the model's structure (a nice summary table)model.summary()
```

## Cell 6 – Add Dense LayersWe can make our model "deeper" (smarter, but also more complex) by adding more Dense layers.Here we go from 1 hidden layer to 2 hidden layers:- First hidden layer: 16 neurons- Second hidden layer: 8 neurons- Output layer: 1 neuron> 🧠 Think of each layer as a filter that refines the information a little more before passing it to the next layer — like sketching a rough drawing, then adding details, then adding final touches.

```python
# A deeper model with two hidden layers instead of onemodel = keras.Sequential([    keras.Input(shape=(1,)),    keras.layers.Dense(16, activation="relu"),  # 1st hidden layer: 16 neurons    keras.layers.Dense(8, activation="relu"),   # 2nd hidden layer: 8 neurons    keras.layers.Dense(1)                       # output layer: 1 number])model.summary()
```

## Cell 7 – Compile ModelBefore training, we must **compile** the model. This step tells Keras *how* it should learn.- `optimizer="adam"` → the strategy the model uses to improve itself after each mistake. "Adam" is a popular, reliable choice — think of it as the model's "study method."- `loss="mse"` → **Mean Squared Error**. This measures *how wrong* the model's predictions are. Lower is better. It's like a "score" for mistakes.- `metrics=["mae"]` → **Mean Absolute Error**, an easy-to-understand extra measurement we track (but don't directly train on) to see how far off, on average, our predictions are.

```python
# Configure how the model will learnmodel.compile(    optimizer="adam",   # how the model updates itself to improve    loss="mse",          # how we measure mistakes during training    metrics=["mae"]      # an extra, easy-to-read measurement of error)print("Model compiled successfully!")
```

## Cell 8 – Train ModelNow the fun part: **training**! This is where the model actually looks at our `X` and `y` data and tries to learn the pattern (y = 2x).- `epochs=100` → the model will go through the *entire* dataset 100 times. Each pass is like a "study session" where it gets a little better.- `verbose=1` → shows a progress bar and the loss value after every epoch, so we can watch it improve.> ⏳ Don't worry if the loss number looks big or small — what matters is that it goes *down* over time.

```python
# Train the model on our datahistory = model.fit(    X,    y,    epochs=100,   # go through the data 100 times    verbose=1     # show progress while training)
```

## Cell 9 – Training HistoryWhile training, Keras keeps a "diary" of how the loss and metrics changed after every epoch. This diary is stored in `history.history`.We can look at it to see exactly how the model's error decreased over time.

```python
# Print the training "diary" — loss and mae recorded at every epochprint(history.history)
```

## Cell 10 – Plot Loss CurveNumbers in a list are hard to read. Let's turn them into a **graph** instead!This chart shows the loss value at each epoch. A healthy training process usually shows the loss line dropping quickly at first, then flattening out as the model gets close to the correct pattern.

```python
# Plot how the loss changed during trainingplt.plot(history.history["loss"])plt.title("Training Loss")plt.xlabel("Epoch")     # how many times the model has seen the full datasetplt.ylabel("Loss")      # how wrong the model's predictions wereplt.show()
```

## Cell 11 – PredictTime to test our model! We'll give it a brand-new input it has never seen before: **6**.Since the real pattern is `y = 2x`, we're hoping the model predicts something close to **12**.

```python
# Ask the model to predict the output for a new input (6)prediction = model.predict(np.array([[6]], dtype=np.float32))print("Predicted value for input 6:", prediction)
```

## Cell 12 – Evaluate Model`evaluate()` checks how well the model performs on data (in this case, the same data it trained on) by calculating the loss and metrics we chose earlier (`mse` and `mae`).This gives us a final "report card" for our model.

```python
# Evaluate the model's performance on our datasetloss, mae = model.evaluate(X, y)print("Loss:", loss)print("MAE:", mae)
```

## Cell 16 – Dropout LayerAs models get bigger, they can sometimes "memorize" the training data instead of truly learning the pattern. This is called **overfitting** — like a student who memorizes answers instead of understanding the topic, and then fails on a slightly different question.**Dropout** helps prevent this. During training, it randomly "turns off" a percentage of neurons in each pass. This forces the model to not rely too heavily on any single neuron, making it generalize better.- `Dropout(0.3)` → randomly turns off 30% of the neurons from the previous layer during each training step.

```python
# A model that uses Dropout to help prevent overfittingmodel = keras.Sequential([    keras.Input(shape=(10,)),    keras.layers.Dense(32, activation="relu"),    keras.layers.Dropout(0.3),   # randomly turns off 30% of neurons while training    keras.layers.Dense(1)])model.summary()
```

## Cell 17 – Flatten LayerSome data, like images, comes in a grid shape (for example, 28 x 28 pixels for a small image). But Dense layers expect a simple, flat list of numbers, not a grid.The **Flatten** layer converts a multi-dimensional input (like a 28x28 image) into a single long 1D list (28 × 28 = 784 numbers) so it can be fed into Dense layers.This example builds a model that could classify images into 10 categories (like digits 0–9), using `softmax` activation in the final layer to output probabilities for each category.

```python
# A model for image-like data (28x28 grid), flattened before the Dense layersmodel = keras.Sequential([    keras.Input(shape=(28, 28)),    keras.layers.Flatten(),                      # turns the 28x28 grid into a flat list of 784 numbers    keras.layers.Dense(128, activation="relu"),    keras.layers.Dense(10, activation="softmax")  # 10 outputs = probabilities for 10 categories])model.summary()
```

## Cell 18 – Early StoppingSometimes if we train for too many epochs, the model starts overfitting instead of improving. **EarlyStopping** is a safety feature that automatically stops training once the model stops getting better.- `monitor="loss"` → keep an eye on the loss value.- `patience=5` → if the loss doesn't improve for 5 epochs in a row, stop training early — no need to keep going if nothing is improving!

```python
# Create an EarlyStopping callback (a helper that watches training and can stop it early)callback = keras.callbacks.EarlyStopping(    monitor="loss",   # what to watch    patience=5        # how many epochs to wait for improvement before stopping)print(callback)
```

## Cell 19 – Model Checkpoint**ModelCheckpoint** automatically saves a copy of the model during training — but only when it's the *best* version seen so far (based on the metric it's tracking). This way, even if training goes wrong later, you still have your best model saved!- `"best_model.keras"` → the filename where the best model will be saved.- `save_best_only=True` → only save when the model improves; don't waste storage on worse versions.

```python
# Create a ModelCheckpoint callback to automatically save the best version of the modelcheckpoint = keras.callbacks.ModelCheckpoint(    "best_model.keras",    save_best_only=True   # only keep the best model seen during training)print(checkpoint)
```

## Cell 20 – TensorBoard Callback**TensorBoard** is a visual dashboard tool that lets you watch your model's training progress in real time through graphs and charts — super helpful for understanding what's happening "under the hood."- `log_dir="logs"` → the folder where training logs will be saved, which TensorBoard then reads to draw its charts.

```python
# Create a TensorBoard callback that will log training data to the "logs" foldertensorboard = keras.callbacks.TensorBoard(log_dir="logs")print(tensorboard)
```
